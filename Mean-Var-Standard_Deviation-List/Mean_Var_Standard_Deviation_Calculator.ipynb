<h1><b>Mean-Variance-Standard Deviation Calculator</b></h1>

---
<h2><b>Project Overview</b></h2>
<p style="text-align: justify;">This notebook implements a function that performs statistical calculations on a list of 9 numbers. The function reshapes the input into a 3×3 matrix and computes various statistical measures across rows, columns, and the entire flattened array.</p>

---

<h2><b>Problem Statement</b></h2>
<p>Create a function named <code>calculate()</code> that:</p>
<ul>
  <li>Accepts a <b>list of 9 numbers</b> as input</li>
  <li>Converts it into a <b>3×3 NumPy array</b></li>
  <li>Returns a dictionary containing:
  <ul>
    <li><b>Mean</b> (average)</li>
    <li><b>Variance</b> (spread of data)</li>
    <li><b>Standard Deviation</b> (average distance from mean)</li>
    <li><b>Max</b> (maximum value)</li>
    <li><b>Min</b> (minimum value)</li>
    <li><b>Sum</b> (total)</li>
  </ul>
  </li>
</ul>
<p>For each statistic, the function should calculate:</p>
<ol>
  <li><b>Axis 0</b> (columns) → list of 3 values</li>
  <li><b>Axis 1</b> (rows) → list of 3 values</li>
  <li><b>Flattened</b> (entire array) → single float</li>
</ol>

---
<h2><b>Requirement</b></h2>
<p>Make sure the following library has installed:</p>


```python
import numpy as np
```



---

<h2><b>Function Definition</b></h2>
<p>Complete the <code>calculate()</code> skeleton below. The function should:</p>
<ol>
<li>Validate that input is a <b>list</b></li>
<li>Validate that list contains <b>exactly 9 numbers</b></li>
<li>Reshape the list into a <b>3×3 matrix</b></li>
<li>Return a dictionary with the required statistics</li>
</ol>

```python
def calculate(lst):
    if type(lst)!=list:
        raise TypeError("Please enter a list.")
    elif len(lst)!=9:
        raise ValueError("List must contain nine numbers.")
    numlist = np.reshape(lst,(3,3))
    calculation = {"mean": [list(np.mean(numlist,axis=0)),list(np.mean(numlist,axis=1)),float(np.mean(numlist))],
                   "variance":[list(np.var(numlist,axis=0)),list(np.var(numlist,axis=1)),float(np.var(numlist))],
                   "standard deviation":[list(np.std(numlist,axis=0)),list(np.std(numlist,axis=1)),float(np.std(numlist))],
                   "max":[list(np.max(numlist,axis=0)),list(np.max(numlist,axis=1)),float(np.max(numlist))],
                   "min":[list(np.min(numlist,axis=0)),list(np.min(numlist,axis=1)),float(np.min(numlist))],
                   "sum":[list(np.sum(numlist,axis=0)),list(np.sum(numlist,axis=1)),float(np.sum(numlist))]}

    return calculation
```
